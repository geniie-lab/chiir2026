## How to mount Google Drive

See [https://github.com/googlecolab/colab-vscode/wiki/User-Guide#working-with-files](https://github.com/googlecolab/colab-vscode/wiki/User-Guide#working-with-files)

- Click `Colab` button at the top of the notebook
- Select `Mount Google Drive to Server`
- Execute the generated cell  and go through yet another series of authentication
    - Connect to Google Drive, Open, (browser) Continue, Continue, (VSCode) Continue, ...
